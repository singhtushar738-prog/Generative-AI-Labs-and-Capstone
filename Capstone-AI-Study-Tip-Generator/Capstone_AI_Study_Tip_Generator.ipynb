```python
import torch, re
from transformers import AutoTokenizer, AutoModelForCausalLM, set_seed
import gradio as gr

device="cuda" if torch.cuda.is_available() else "cpu"
MODEL_NAME="gpt2"
tokenizer=AutoTokenizer.from_pretrained(MODEL_NAME)
model=AutoModelForCausalLM.from_pretrained(MODEL_NAME).to(device)
tokenizer.pad_token=tokenizer.eos_token
set_seed(42)
print("Model loaded on",device)
```

## 1. Define the generation function

```python
def generate_completion(prompt, max_new_tokens=180):
    inputs=tokenizer(prompt,return_tensors="pt").to(device)
    with torch.no_grad():
        ids=model.generate(
            **inputs,max_new_tokens=max_new_tokens,
            do_sample=True,temperature=0.75,top_p=0.9,
            repetition_penalty=1.1,
            pad_token_id=tokenizer.eos_token_id
        )
    full=tokenizer.decode(ids[0],skip_special_tokens=True)
    return full[len(prompt):].strip()

def study_assistant(topic, level):
    topic=(topic or "").strip()
    if len(topic)<3:
        return "Please enter a study topic containing at least 3 characters."
    prompt=(
        f"You are a helpful study assistant for a {level} student. "
        f"Create clear, accurate study guidance about: {topic}. "
        "Give a short explanation, three revision tips, and two practice questions. "
        "Use simple language.\nStudy guide:\n"
    )
    answer=generate_completion(prompt, max_new_tokens=180)
    return answer if answer else "The model did not generate a response. Please try again."
```

## 2. Test the application function

```python
sample=study_assistant("Neural Networks","beginner")
print(sample)
```

## 3. Create the Gradio user interface

```python
demo=gr.Interface(
    fn=study_assistant,
    inputs=[
        gr.Textbox(label="Study topic",placeholder="e.g. Generative Adversarial Networks"),
        gr.Dropdown(["beginner","intermediate","advanced"],value="beginner",label="Learning level")
    ],
    outputs=gr.Textbox(label="Generated study guide",lines=12),
    title="AI Study Tip Generator",
    description="Enter a topic and learning level to generate a short study guide using GPT-2. Verify facts with trusted learning resources."
)
demo.launch(share=True)
```

## 4. Example test cases

```python
test_topics=[
    ("Overfitting in machine learning","beginner"),
    ("Transformer attention mechanism","intermediate"),
    ("Variational Autoencoders","advanced")
]
for t,lvl in test_topics:
    print("\n"+"="*70)
    print("TOPIC:",t,"| LEVEL:",lvl)
    print(study_assistant(t,lvl))
```
